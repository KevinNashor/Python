```python
nombres = ['Aaron','Jose','Robin','Airam']
print(nombres[2])
print(type(nombres))
del nombres[2]
print(nombres)
nombres.insert(2,'Robin')
print(nombres)
print('Para ver la posicion donde se encuentra un valo usamos index, Jose se encuentra en la posicion:',nombres.index('Jose'))
print(nombres)
print('Para ver los elementos de la posicion 1 a la 2, pongo un 3 al final!',nombres[1:3])
nombres.append('Mana')
print(nombres)
print('para ver los elementos de 2 en 2:',nombres[::2])
print('para ver los elementos en orden invertido uso -1',nombres[::-1])
# para listar uno a uno los elementos de la lista
for i in nombres:
    print(i)
```
